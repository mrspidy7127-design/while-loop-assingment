## Part 1: While Loop


**Q1.** Ek while loop banayein jo 'Hello' ko 3 martaba print kare.

```python
# Apna code yahan likhein
total = 0

while total < 3:
    print("Hello")
    total+=1
```

**Q2.** 10 se 1 tak ulti ginti (reverse counting) print karne ka while loop likhein.

```python
# Apna code yahan likhein

total = 10

while total >=1:
    print(total)
    total= total-1
    
```

**Q9.** String "Ali" ke har harf (letter) ko while loop aur index ka istemal karke alag line mein print karein.

**Q10.** Ek while loop mein `continue` ka istemal karein taake wo number 3 ko skip kar ke 1 se 5 tak print kare.

```python
# Apna code yahan likhein
i = 0
while i < 5:
    i+=1

    if i == 3:
        continue
    print(i)

```

**Q4.** Ek function `find_maximum(a, b, c)` banayein jo 3 numbers mein se sabse bada number bataye.

```python
# Apna code yahan likhein

def find_maximum(a, b, c):
    if a >= b and a >= c:
        print(f"{a} is greater")
    elif b >= a and b >= c:
        print(f"{b} is greater")
    else:
        print(f"{c} is greater")
```

```python
(find_maximum(5, 8, 9))
```

**Q6.** Ek function `celsius_to_fahrenheit(c)` banayein jo temperature convert kare.

```python
# Apna code yahan likhein

def celsius_to_fahrenheit(c):
    fahrenheit = (c * 9/5) + 32
    return fahrenheit
```

**Q7.** Ek function `is_palindrome(word)` banayein jo check kare ke string palindrome hai ya nahi.

```python
# Apna code yahan likhein
def is_palindrome(word):
    if word == word[: :-1]:
        print(f"'{word}' ek palindrome ha!")
    else:
        print(f"'{word}' palindrome nhi ha!")
```

**Q9.** Ek function `calculate_factorial(n)` banayein jo number ka factorial return kare.

```python
# Apna code yahan likhein
def calculate_factorial(n):
    tot = 1

    while n > 0:
        result = tot *n
        tot = result
        n-=1
    return tot
```

```python
(calculate_factorial(5))
```

**Q10.** Ek function `count_vowels(text)` banayein jo string mein vowels (a, e, i, o, u) ki tadad bataye.

```python
# Apna code yahan likhein

def count_vowels(text):
    total = 0
    vowels = "aeiou"
    text = text.lower()
    for s in text:
        if s in vowels:
            total+=1
    return total   
```

```python
(count_vowels("Apple"))
```
